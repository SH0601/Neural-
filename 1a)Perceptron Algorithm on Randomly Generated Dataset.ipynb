```python

import numpy as np
from sklearn import datasets
import matplotlib.pyplot as plt

X, y = datasets.make_blobs(n_samples=150, n_features=2,
                           centers=2, cluster_std=1,
                           random_state=2)

plt.figure(figsize=(8, 8))
plt.plot(X[y == 0, 0], X[y == 0, 1], 'r^')
plt.plot(X[y == 1, 0], X[y == 1, 1], 'bs')
plt.xlabel("feature 1")
plt.ylabel("feature 2")
plt.title("Random Classification Data with 2 classes")
plt.show()

def step_func(z):
    return 1.0 if z > 0 else 0.0

def perceptron(X, y, lr, epochs):
    theta = np.zeros((X.shape[1] + 1, 1))
    for _ in range(epochs):
        for i, x in enumerate(X):
            x = np.insert(x, 0, 1).reshape(-1, 1)
            pred = step_func((x.T @ theta).item())
            if pred != y[i]:
                theta += lr * (y[i] - pred) * x
    return theta

theta = perceptron(X, y, 0.5, 100)

plt.figure(figsize=(8, 8))
plt.plot(X[y == 0, 0], X[y == 0, 1], 'r^')
plt.plot(X[y == 1, 0], X[y == 1, 1], 'bs')

x1 = [X[:, 0].min(), X[:, 0].max()]
m = -theta[1, 0] / theta[2, 0]
c = -theta[0, 0] / theta[2, 0]

plt.plot(x1, m * np.array(x1) + c, 'y-')
plt.xlabel("feature 1")
plt.ylabel("feature 2")
plt.title("Perceptron Algorithm")
plt.show()

```
