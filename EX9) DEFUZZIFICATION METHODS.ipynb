```python
import numpy as np
import matplotlib.pyplot as plt
import skfuzzy as f

x = np.arange(0, 5.05, 0.01)
m = f.trapmf(x, [2, 2.5, 3, 4.5])

a = ['centroid', 'bisector', 'mom', 'som', 'lom']
v = [f.defuzz(x, m, i) for i in a]
n = ['centroid', 'bisector', 'mom',
     'som', 'lom']

plt.plot(x, m, 'k')

for i, c, name in zip(v, 'rbgcm', n):
    plt.vlines(i, 0, np.interp(i, x, m), color=c, label=name)

plt.xlim(0, 5)
plt.ylim(-0.1, 1.1)
plt.xlabel('Universe variable (arb)')
plt.ylabel('Fuzzy membership')
plt.legend(loc=2)
plt.show()
```
