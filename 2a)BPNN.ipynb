```python

import numpy as np

X = np.array([[2,9],[1,5],[3,6]], float)
y = np.array([[92],[86],[89]], float)

X = X / X.max(axis=0)
y = y / 100

class Neural_Network:
    def __init__(self):
        self.W1 = np.random.randn(2,3)
        self.W2 = np.random.randn(3,1)

    def sigmoid(self, s):
        return 1 / (1 + np.exp(-s))

    def forward(self, X):
        self.z = X @ self.W1
        self.z2 = self.sigmoid(self.z)
        self.z3 = self.z2 @ self.W2
        return self.sigmoid(self.z3)

    def train(self, X, y, o):
        e = y - o
        d = e * o * (1-o)
        h = (d @ self.W2.T) * self.z2 * (1-self.z2)
        self.W1 += X.T @ h
        self.W2 += self.z2.T @ d

NN = Neural_Network()

for i in range(1000):
    o = NN.forward(X)
    print("Input:\n", X)
    print("Actual Output:\n", y)
    print("Predicted Output:\n", o)
    print("Loss:\n", np.mean((y-o)**2))
    print()
    NN.train(X, y, o)
```
