```python
import numpy as np

class Perceptron:
    def __init__(self):
        self.weights = np.zeros((2, 1))
        self.bias = 0

    def predict(self, x):
        return np.where(np.dot(x, self.weights) + self.bias > 0, 1, 0)

    def fit(self, X, y, epoch=100, lr=0.1):
        for _ in range(epoch):
            for xi, yi in zip(X, y):
                p = self.predict(xi).squeeze()
                e = yi - p
                self.weights += lr * e * xi.reshape(-1, 1)
                self.bias += lr * e

    def score(self, X, y):
        return np.mean(self.predict(X).squeeze() == y)


X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([0, 1, 1, 1])

p = Perceptron()
print("OR Gate: ")
for x, yi in zip(X, y):
    print(x, yi)

print("Initial Weights:")
print(p.weights, p.bias)

p.fit(X, y, epoch=3)

print("Trained Weights:")
print(p.weights, p.bias)

print("OR Gate(predicted): ")
for x in X:
    print(x, p.predict(x))

print("Accuracy:", p.score(X, y))
```
