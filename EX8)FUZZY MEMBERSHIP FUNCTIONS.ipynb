```python
import numpy as np
import skfuzzy as f
import matplotlib.pyplot as plt

x = np.arange(0, 11, 1)
x1 = np.arange(0, 10, 0.1)

y1 = f.trimf(x, [1, 3, 5])
y2 = f.trapmf(x, [1, 3, 5, 7])
y3 = f.gbellmf(x1, 2, 4, 5)
y4 = f.sigmf(x, 5, 1)
y5 = f.gaussmf(x, 5, 1)
y6 = np.clip((x-2)/6, 0, 1)

y = [y1, y2, y3, y4, y5, y6]

t = ["Triangular", "Trapezoidal", "Bell-shaped",
     "Sigmoid", "Gaussian", "Linear"]

fig, ax = plt.subplots(6, 1, figsize=(10, 10))

for i in range(6):
    ax[i].plot(x1 if i == 2 else x, y[i], linewidth=2)
    ax[i].set_title(t[i], fontsize=14)
    

plt.tight_layout()
plt.show()
```
