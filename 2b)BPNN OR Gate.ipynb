```python
import numpy as np

X = np.array([[0,0],[0,1],[1,0],[1,1]], float)
Y = np.array([[0],[1],[1],[1]], float)

print("X:",X)
print("Y:",Y)
print("X shape:",X.shape)
print("Y shape:",Y.shape)

class NeuralNetwork:

    def __init__(self):
        self.w1=np.array([[0.6,-0.3],[-0.1,0.4]])
        self.b1=np.array([[0.3,0.5]])
        self.w2=np.array([[0.4],[0.1]])
        self.b2=np.array([[-0.2]])

    def sigmoid(self,s):
        return 1/(1+np.exp(-s))

    def sigmoidprime(self,s):
        return s*(1-s)

    def forward(self,X):
        self.z=np.dot(X,self.w1)+self.b1
        print("Z:",self.z)

        self.z2=self.sigmoid(self.z)
        print("Z2:",self.z2)

        self.z3=np.dot(self.z2,self.w2)+self.b2
        print("Z3:",self.z3)

        o=self.sigmoid(self.z3)
        print("Output:",o)
        return o

    def backward(self,X,Y,o):
        error=Y-o
        print("Error:",error)

        delta=error*self.sigmoidprime(o)
        print("Delta:",delta)

        z_error=delta.dot(self.w2.T)
        z_delta=z_error*self.sigmoidprime(self.z2)

        self.w1+=X.T.dot(z_delta)*0.25
        print("W1:",self.w1)

        self.w2+=self.z2.T.dot(delta)*0.25
        print("W2:",self.w2)

    def train(self,X,Y):
        o=self.forward(X)
        self.backward(X,Y,o)

nn=NeuralNetwork()
nn.train(X,Y)

print("Input:",X)
print("Actual Output:",Y)
print("Loss:",np.mean(np.square(Y-nn.forward(X))))
```
