```python
import numpy as np
import pandas as pd
from scipy.spatial.distance import cdist

X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([0,1,1,0])
weight = np.array([-1,1,1,-1])

centers = X.copy()
print("Centers:\n", centers)

std_dev = 1

rbf_x = np.around(np.exp(-cdist(X, centers, 'sqeuclidean') / 2), 1)

pred = np.dot(rbf_x, weight)
step = np.where(pred > 0, 1, 0)

df = pd.DataFrame({
    'Input': list(map(tuple, X)),
    'RBF Activation': list(map(tuple, rbf_x)),
    'Predicted Value': pred,
    'Predicted Output': step
})

print("\nDataFrame:\n", df)
```
