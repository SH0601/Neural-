```python
import numpy as np

a = np.array([1.0, 0.4, 0.2])
b = np.array([0.2, 0.43, 0.44])
c = np.array([0.2, 0.93, 0.24])

e = np.ones(3)
o = np.zeros(3)

print("A:", a)
print("B:", b)
print("C:", c)

print("\nUnion:", np.maximum(a, b))
print("Intersection:", np.minimum(a, b))

print("\nCommutative Property:")
print(np.maximum(a, b), np.maximum(b, a))
print(np.minimum(a, b), np.minimum(b, a))

print("\nAssociative Property:")
print(np.maximum(np.maximum(a, b), c))
print(np.maximum(a, np.maximum(b, c)))
print(np.minimum(np.minimum(a, b), c))
print(np.minimum(a, np.minimum(b, c)))

print("\nDistributive Property:")
print(np.maximum(a, np.minimum(b, c)))
print(np.minimum(np.maximum(a, b), np.maximum(a, c)))
print(np.minimum(a, np.maximum(b, c)))
print(np.maximum(np.minimum(a, b), np.minimum(a, c)))

print("\nAbsorption Property:")
print(np.maximum(a, np.minimum(a, b)))
print(np.minimum(a, np.maximum(a, b)))

print("\nIdempotent Law:")
print(np.maximum(a, a))
print(np.minimum(a, a))

print("\nInvolution Property:")
print(1 - (1 - a))

print("\nDe Morgan's Law:")
print(1 - np.maximum(a, b))
print(np.minimum(1-a, 1-b))
print(1 - np.minimum(a, b))
print(np.maximum(1-a, 1-b))

print("\nIdentity Property:")
print(np.maximum(a, o))
print(np.minimum(a, e))

print("\nExcluded Middle:")
print(np.maximum(a, 1-a))
print(np.minimum(a, 1-a))
```
